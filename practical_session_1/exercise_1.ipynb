But python alone is not useful for mathematics, that's why we have libraries, and we are going to import them ;)

```python
import numpy as np
a=np.exp(-2.0)
print(a)
```

Now we are going to make a list of data

```python
x=np.arange(0.0,2.0,0.1)
print(x)
```

if we want it to end at 2.0, we need to punt inside the arange 2.1 instead of 2.0 as the end

```python
x=np.arange(0.0,2.1,0.1)
print(x)
```

```python
y=np.exp(x)
print(y)
```

the library to make plots is matplotlib and thats what we are going to use

```python
import matplotlib.pyplot as plt
plt.plot(x,y,'gv')
plt.title('Exponencial')
plt.show()
```
