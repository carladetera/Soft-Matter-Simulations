## Lists

```python
# This is a list
energy_kcal = [-13.4, -2.7, 5.4, 42.1]
# I can determine its length
energy_length = len(energy_kcal)

# print the list length
print('The length of this list is', energy_length)
```

```python
# Print the first element of the list
print(energy_kcal[0])
```

```python
# Convert the second list element to kilojoules.
energy_kilojoules = energy_kcal[1] * 4.184
print(energy_kilojoules, 'kJ')
```

## Repeating an operation many times: for loops

```python
for number in energy_kcal:
    kJ = number * 4.184
    print(kJ)
```

```python
energy_kJ=[] #ha d'existir abans no la mencionem
for number in energy_kcal:
    kJ = number * 4.184
    energy_kJ.append(kJ)

print(energy_kJ)
```

## Making choices: logic Statements

```python
negative_energy_kJ = []
positive_energy_kJ=[]

for number in energy_kJ:
    if number < 0:
        negative_energy_kJ.append(number)
    elif number >0:
        positive_energy_kJ.append(number)
        

print('negative list:', negative_energy_kJ,'positive list:', positive_energy_kJ)
```

```python
negative_numbers = []
for number in energy_kJ:
    if number < 0 or number == 0:
        negative_numbers.append(number)

print(negative_numbers)
```
